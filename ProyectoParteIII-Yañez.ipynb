```python
# Importación de librerías específicas para el modelado
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
# Se replican los pasos de limpieza de la Parte I:
df = pd.read_csv("ScreenTime vs MentalWellness.csv")
if 'Unnamed: 15' in df.columns:
    df.drop(columns=["Unnamed: 15"], inplace=True)
Q1 = df["screen_time_hours"].quantile(0.25)
Q3 = df["screen_time_hours"].quantile(0.75)
IQR = Q3 - Q1
lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR
df = df[(df["screen_time_hours"] >= lim_inf) & (df["screen_time_hours"] <= lim_sup)].copy()
```

```python
# A. Definir Target (Y) y Features (X)
target = 'mental_wellness_index_0_100'
Y = df[target]
features_to_encode = df.drop(columns=[target, 'user_id'])

# B. Feature Selection (Codificación Categórica con One-Hot Encoding)
# Se aplica OHE para convertir 'gender', 'work_mode' y 'occupation' en variables binarias numéricas.
X_encoded = pd.get_dummies(features_to_encode, drop_first=True)

# C. División de Datos (Train-Test Split)
# Se utiliza el 80% para entrenar y el 20% para la prueba final
X_train, X_test, Y_train, Y_test = train_test_split(
    X_encoded, Y, test_size=0.2, random_state=42
)

# D. Estandarización de las Features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Feature Selection y Preparación


1.  **Codificación Categórica (One-Hot Encoding):** El algoritmo de Regresión Lineal solo trabaja con valores numéricos. Por lo tanto, se aplicó **One-Hot Encoding** a las variables categóricas (`gender`, `work_mode`, `occupation`), transformando cada categoría en una nueva columna binaria (0 o 1). Este método aumenta la cantidad de *features* de $\approx 11$ a **18** columnas, preparándolas para el modelo.
2.  **División de Datos (Train-Test Split):** Se dividió el dataset en conjuntos de entrenamiento (80%) y prueba (20%) para **validar la generalización del modelo** y asegurar que no haya un sobreajuste (*overfitting*) a los datos que ya conoce.
3.  **Estandarización de Features:** Se escalaron los datos con `StandardScaler` para asegurar que todas las variables (como `age` vs. `stress_level_0_10`) tengan una media de cero y desviación estándar de uno. Esto evita que las *features* con valores más grandes (`exercise_minutes_per_week`) dominen el cálculo de los coeficientes del modelo.

```python
# Elegir el Algoritmo: Regresión Lineal
model = LinearRegression()

# Entrenar el modelo con los datos de entrenamiento escalados
model.fit(X_train_scaled, Y_train)

# Realizar predicciones con el conjunto de prueba escalado
Y_pred = model.predict(X_test_scaled)

# --- Cálculo de Métricas para la Validación del Modelo ---

# R² Score (Coeficiente de Determinación)
r2 = r2_score(Y_test, Y_pred)

# MAE (Error Absoluto Medio)
mae = mean_absolute_error(Y_test, Y_pred)

# RMSE (Raíz del Error Cuadrático Medio)
rmse = np.sqrt(mean_squared_error(Y_test, Y_pred))

print("--- Métricas de Desempeño del Modelo de Regresión Lineal ---")
print(f"R² Score: {r2:.4f}")
print(f"MAE: {mae:.4f} puntos")
print(f"RMSE: {rmse:.4f} puntos")

# Extracción de Coeficientes para el análisis final
coefficients = pd.Series(model.coef_, index=X_encoded.columns).sort_values(ascending=False)

print("\n--- Coeficientes Principales (Impacto en Bienestar) ---")
print(coefficients.head(5).to_markdown(numalign="left", stralign="left"))
print(coefficients.tail(5).to_markdown(numalign="left", stralign="left"))
```

```python
# ANÁLISIS VISUAL DE RESIDUOS

residuos = Y_test - Y_pred

plt.figure(figsize=(8, 5))
plt.scatter(Y_pred, residuos, alpha=0.6)
plt.hlines(y=0, xmin=Y_pred.min(), xmax=Y_pred.max(), color='red', linestyle='--')
plt.title('Análisis de Residuos')
plt.xlabel('Valores Predichos (Bienestar)')
plt.ylabel('Residuos (Error: Real - Predicho)')
plt.grid(True)
plt.show()

# Interpretación: Si los puntos están dispersos y centrados alrededor de la línea roja (y=0),
# el modelo cumple con los supuestos de linealidad.
```
