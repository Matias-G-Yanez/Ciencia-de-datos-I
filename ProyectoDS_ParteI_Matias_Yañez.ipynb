Abstract

El presente proyecto tiene como objetivo analizar la relación entre el uso de pantallas (screen time) y diferentes indicadores de bienestar mental en una muestra de 400 participantes. Este dataset recopila información variada sobre características demográficas (edad, género, ocupación, modalidad de trabajo), hábitos diarios (horas frente a pantallas por trabajo y ocio, horas de sueño, minutos de ejercicio, interacción social) y variables psicológicas clave (nivel de estrés, productividad percibida, calidad del sueño, índice de bienestar mental).

La elección de este conjunto de datos responde a la relevancia creciente del tema en la vida contemporánea. El aumento del tiempo frente a pantallas, tanto en contextos laborales como recreativos, ha generado un interés académico y social en comprender sus efectos sobre la salud mental. Factores como la productividad, la calidad del sueño y los niveles de estrés están íntimamente ligados a la manera en que las personas utilizan la tecnología, lo cual convierte a este dataset en una fuente valiosa para explorar tendencias y patrones significativos.

Para abordar estas preguntas se aplicarán técnicas de análisis exploratorio de datos, incluyendo visualizaciones univariadas (histogramas, boxplots), bivariadas (diagramas de dispersión, gráficos de barras comparativos) y multivariadas (gráficos con agrupación por variables demográficas).
```python
# Boxplots para detectar valores atípicos
plt.figure(figsize=(8,5))
sns.boxplot(x=df["screen_time_hours"])
plt.title("Boxplot de horas de pantalla")
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(x=df["stress_level_0_10"])
plt.title("Boxplot de nivel de estrés")
plt.show()
```

```python
Q1 = df["screen_time_hours"].quantile(0.25)
Q3 = df["screen_time_hours"].quantile(0.75)
IQR = Q3 - Q1

lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

print("Límite inferior:", lim_inf)
print("Límite superior:", lim_sup)

outliers = df[(df["screen_time_hours"] < lim_inf) | (df["screen_time_hours"] > lim_sup)]
print("Cantidad de outliers en screen_time_hours:", len(outliers))
```

```python
df = df[(df["screen_time_hours"] >= lim_inf) & (df["screen_time_hours"] <= lim_sup)]
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["screen_time_hours"])
plt.title("Boxplot de horas de pantalla")
plt.xlabel("Horas de pantalla diarias")
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(x=df["stress_level_0_10"])
plt.title("Boxplot de nivel de estrés")
plt.xlabel("Nivel de estrés (0 a 10)")
plt.show()

# En el caso de las horas de pantalla, se detectaron 8 outliers con valores inferiores a 2.7 horas y superiores a 15.2 horas diarias. Al eliminarlos, la distribución se centra mejor en los hábitos típicos de la mayoría de los usuarios, entre 7 y 12 horas. En cuanto al nivel de estrés, los outliers se ubican en los valores más bajos (0 a 2), lo que indica que son pocos los usuarios que reportan niveles de estrés casi nulos
```

**Hipótesis 1 – Relación entre screentime y estrés**

Se plantea que el aumento en las horas de uso de pantallas tiene un impacto negativo en el bienestar emocional, elevando los niveles de estrés percibido. La lógica detrás de esta hipótesis es que una exposición prolongada a pantallas, tanto en el trabajo como fuera de él, puede asociarse con fatiga mental, reducción de descanso y sobrecarga cognitiva. Por lo tanto, se espera observar una relación positiva entre el tiempo de screentime y el nivel de estrés reportado por los participantes.


---

```python
# Relación entre horas de pantalla y nivel de estrés
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df,
    x="screen_time_hours",
    y="stress_level_0_10",
    hue="gender",
    palette="Set2",
    alpha=0.7
)

# Línea de tendencia
sns.regplot(
    data=df,
    x="screen_time_hours",
    y="stress_level_0_10",
    scatter=False,
    color="red"
)

plt.title("Relación entre horas de pantalla y nivel de estrés")
plt.xlabel("Horas de pantalla (diarias)")
plt.ylabel("Nivel de estrés (0-10)")
plt.show()


# El gráfico muestra cómo varía el nivel de estrés en función del tiempo de pantalla. La línea roja tiene pendiente positiva, lo cual respalda la hipótesis de que más screentime se asocia con mayor estrés.
```



---


**Hipótesis 2 – El rol del ejercicio físico como factor protector**

El ejercicio regular es reconocido como una actividad beneficiosa para la salud mental. En este contexto, la hipótesis sostiene que los individuos que realizan más minutos de actividad física semanal muestran un mayor índice de bienestar mental, incluso en presencia de altos niveles de screentime. En otras palabras, la práctica de ejercicio podría actuar como un factor moderador que compensa los efectos negativos de la exposición a pantallas, favoreciendo un mejor balance entre vida digital y salud mental.

---



```python
# Relación entre ejercicio semanal y bienestar mental, coloreado por screentime
scatter = plt.scatter(
    df["exercise_minutes_per_week"],
    df["mental_wellness_index_0_100"],
    c=df["screen_time_hours"],
    cmap="coolwarm",
    alpha=0.7
)

plt.colorbar(scatter).set_label("Horas de pantalla")
plt.title("Ejercicio semanal y bienestar mental")
plt.xlabel("Minutos de ejercicio por semana")
plt.ylabel("Índice de bienestar mental (0-100)")
plt.show()


# Se observa que personas con más ejercicio tienden a tener un índice de bienestar más alto, incluso con altos niveles de screentime. Se confirmar la hipótesis del efecto protector del ejercicio.
```



---

**Hipótesis 3 – Diferencias según modalidad de trabajo**

Se propone que la modalidad de trabajo influye en los hábitos digitales y en el nivel de estrés de los participantes. Aquellos que trabajan de forma remota tienden a pasar más horas frente a pantallas que quienes lo hacen de manera presencial o híbrida. En consecuencia, es probable que los trabajadores remotos reporten mayores niveles de estrés, reflejando la carga extra que implica la dependencia tecnológica en su rutina laboral.

---



```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Gráfico violin plot - Comparación del nivel de estrés según modalidad de trabajo

plt.figure(figsize=(8,5))
sns.violinplot(x="work_mode", y="stress_level_0_10", data=df, palette="Set2")
plt.title("Distribución del nivel de estrés según modalidad de trabajo")
plt.xlabel("Modalidad de trabajo")
plt.ylabel("Nivel de estrés (0-10)")
plt.show()

# Se observa que en las tres modalidades hay una alta concentración de valores cercanos al nivel de estrés máximo (9–10). Aunque el estrés alto es predominante en todas las modalidades, el trabajo presencial parece ofrecer un mayor número de casos con niveles de estrés bajos.
```
